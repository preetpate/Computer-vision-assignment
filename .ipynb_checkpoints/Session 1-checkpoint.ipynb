2 — Read and display an image

```python
import cv2
import matplotlib.pyplot as plt

image_path = "cricketer_img.png"  # Replace with your image filename
image = cv2.imread(image_path)

if image is None:
    print(f"Error: Could not load '{image_path}'. Check the filename and folder location.")
else:
    # OpenCV loads images in BGR format; convert to RGB for correct matplotlib colours.
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(8, 6))
    plt.imshow(image_rgb)
    plt.axis("off")
    plt.title("My Image")
    plt.show()
```

3 — Print image shape and use a custom window title

```python
if image is None:
    print("Load an image successfully in Task 2 before running this cell.")
else:
    height, width, channels = image.shape
    print("Image shape (height, width, channels):", image.shape)
    print("Height:", height)
    print("Width:", width)
    print("Channels:", channels)

    # Display inside the Jupyter notebook with a custom title.
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(8, 6))
    plt.imshow(image_rgb)
    plt.axis("off")
    plt.title("Instagram Selfie")  # Change this title if you prefer.
    plt.show()
```

4 — How computer vision is used in an app

Instagram uses computer vision in its face filters to detect a person's face and identify important facial features, such as the eyes, nose, mouth, and outline of the face. The camera image is analysed in real time, and the app uses the detected facial landmarks to place digital effects—such as glasses, makeup, masks, or animations—in the correct positions. As the person moves their head or changes their expression, the system tracks those features and updates the filter so that it follows the face. In this way, computer vision helps Instagram understand what is visible in the camera image and react to it instantly.
