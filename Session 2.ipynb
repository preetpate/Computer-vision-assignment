1 — Resize a selfie to exactly 300 × 300 pixels

```python
import cv2
import matplotlib.pyplot as plt

selfie_path = "cricketer_img.png"  
selfie = cv2.imread(selfie_path)

if selfie is None:
    print(f"Error: Could not read '{selfie_path}'. Check the filename and folder.")
else:
    resized_selfie = cv2.resize(
        selfie,
        (300, 300),
        interpolation=cv2.INTER_AREA
    )

    cv2.imwrite("resized_selfie.jpg", resized_selfie)
    print("Original dimensions (height, width):", selfie.shape[:2])
    print("Resized dimensions (height, width):", resized_selfie.shape[:2])
    print("Saved as resized_selfie.jpg")

    # Display the resized image with correct colours
    plt.figure(figsize=(5, 5))
    plt.imshow(cv2.cvtColor(resized_selfie, cv2.COLOR_BGR2RGB))
    plt.title("Resized Selfie — 300 × 300")
    plt.axis("off")
    plt.show()
```

2 — Display original dimensions and resize to half size

```python
food_path = "cricketer_img.png"  
food = cv2.imread(food_path)

if food is None:
    print(f"Error: Could not read '{food_path}'. Check the filename and folder.")
else:
    original_height, original_width = food.shape[:2]
    print("Original width:", original_width)
    print("Original height:", original_height)

    half_width = original_width // 2
    half_height = original_height // 2

    half_size_food = cv2.resize(
        food,
        (half_width, half_height),
        interpolation=cv2.INTER_AREA
    )

    print("Resized width:", half_size_food.shape[1])
    print("Resized height:", half_size_food.shape[0])

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(food, cv2.COLOR_BGR2RGB))
    plt.title("Original cricketer Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(half_size_food, cv2.COLOR_BGR2RGB))
    plt.title("Half-size cricketer Image")
    plt.axis("off")
    plt.tight_layout()
    plt.show()
```

3 — Split an IPL team logo into R, G, and B channels

```python
logo_path = "cricketer_img.png"  
logo_bgr = cv2.imread(logo_path)

if logo_bgr is None:
    print(f"Error: Could not read '{logo_path}'. Check the filename and folder.")
else:
    logo_rgb = cv2.cvtColor(logo_bgr, cv2.COLOR_BGR2RGB)
    red_channel, green_channel, blue_channel = cv2.split(logo_rgb)

    cv2.imwrite("logo_R.jpg", red_channel)
    cv2.imwrite("logo_G.jpg", green_channel)
    cv2.imwrite("logo_B.jpg", blue_channel)

    print("Saved: logo_R.jpg, logo_G.jpg, logo_B.jpg")

    plt.figure(figsize=(12, 4))

    plt.subplot(1, 3, 1)
    plt.imshow(red_channel, cmap="gray")
    plt.title("Red Channel")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(green_channel, cmap="gray")
    plt.title("Green Channel")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(blue_channel, cmap="gray")
    plt.title("Blue Channel")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```

4 — Convert a product image to grayscale and compare

```python
product_path = "cricketer_img.png"  # Change this to your product image filename
product_bgr = cv2.imread(product_path)

if product_bgr is None:
    print(f"Error: Could not read '{product_path}'. Check the filename and folder.")
else:
    product_rgb = cv2.cvtColor(product_bgr, cv2.COLOR_BGR2RGB)
    product_gray = cv2.cvtColor(product_bgr, cv2.COLOR_BGR2GRAY)

    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(product_rgb)
    plt.title("Original Product Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(product_gray, cmap="gray")
    plt.title("Grayscale Product Image")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```

Task 5 — Why do computer vision apps use grayscale images?

Many computer vision apps use grayscale images because they contain brightness information without the extra complexity of colour, which can reduce processing time and make important edges, shapes, and patterns easier to analyse. For example, Google Drive's document scanning feature** can use grayscale-style processing to improve the readability of scanned text and document edges.
