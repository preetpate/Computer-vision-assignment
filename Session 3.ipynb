```python
import cv2
import matplotlib.pyplot as plt

print("OpenCV version:", cv2.__version__)
```

 1 — Resize a selfie to exactly 300 × 300 pixels

```python
selfie_path = "cricketer_img.png"  
selfie = cv2.imread(selfie_path)

if selfie is None:
    print(f"Error: Could not read '{selfie_path}'. Check the filename and folder.")
else:
    resized_selfie = cv2.resize(selfie, (300, 300), interpolation=cv2.INTER_AREA)
    cv2.imwrite("resized_selfie.jpg", resized_selfie)

    print("Original dimensions (height, width):", selfie.shape[:2])
    print("New dimensions (height, width):", resized_selfie.shape[:2])
    print("Saved as resized_selfie.jpg")

    plt.figure(figsize=(5, 5))
    plt.imshow(cv2.cvtColor(resized_selfie, cv2.COLOR_BGR2RGB))
    plt.title("Resized Selfie — 300 × 300")
    plt.axis("off")
    plt.show()
```

2 — Rotate an IPL team logo 45° clockwise

```python
logo_path = "cricketer_img.png"  
logo = cv2.imread(logo_path)

if logo is None:
    print(f"Error: Could not read '{logo_path}'. Check the filename and folder.")
else:
    height, width = logo.shape[:2]
    center = (width // 2, height // 2)

    rotation_matrix = cv2.getRotationMatrix2D(center, -45, 1.0)
    rotated_logo = cv2.warpAffine(logo, rotation_matrix, (width, height))

    cv2.imwrite("ipl_logo_rotated.png", rotated_logo)
    print("Saved as King_logo_rotated.png")

    plt.figure(figsize=(6, 6))
    plt.imshow(cv2.cvtColor(rotated_logo, cv2.COLOR_BGR2RGB))
    plt.title("King img Rotated 45° Clockwise")
    plt.axis("off")
    plt.show()
```

3 — Crop only the food item using array slicing

```python
food_path = "cricketer_img.png"  
food = cv2.imread(food_path)

if food is None:
    print(f"Error: Could not read '{food_path}'. Check the filename and folder.")
else:
    height, width = food.shape[:2]
    print("Image width:", width, "| Image height:", height)

    x1, y1 = 100, 100
    x2, y2 = min(400, width), min(400, height)

    if x1 < x2 and y1 < y2 and x2 <= width and y2 <= height:
        food_cropped = food[y1:y2, x1:x2]
        cv2.imwrite("food_cropped.jpg", food_cropped)
        print("Saved as food_cropped.jpg")

        plt.figure(figsize=(10, 5))
        plt.subplot(1, 2, 1)
        plt.imshow(cv2.cvtColor(food, cv2.COLOR_BGR2RGB))
        plt.title("Original King Photo")
        plt.axis("off")

        plt.subplot(1, 2, 2)
        plt.imshow(cv2.cvtColor(food_cropped, cv2.COLOR_BGR2RGB))
        plt.title("Cropped King Img")
        plt.axis("off")
        plt.tight_layout()
        plt.show()
    else:
        print("Invalid crop coordinates. Adjust x1, y1, x2, and y2.")
```

Task 4 — Flip an Instagram story screenshot horizontally



```python
story_path = "story.jpg"  
story = cv2.imread(story_path)

if story is None:
    print(f"Error: Could not read '{story_path}'. Check the filename and folder.")
else:
    flipped_story = cv2.flip(story, 1)  
    cv2.imwrite("insta_story_flipped.jpg", flipped_story)
    print("Saved as insta_story_flipped.jpg")

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(story, cv2.COLOR_BGR2RGB))
    plt.title("Original Story")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(flipped_story, cv2.COLOR_BGR2RGB))
    plt.title("Horizontally Flipped Story")
    plt.axis("off")
    plt.tight_layout()
    plt.show()
```

Task 5 — Resize and then flip a Myntra product image vertically



```python
product_path = "myntra_product.jpg"  
product = cv2.imread(product_path)

if product is None:
    print(f"Error: Could not read '{product_path}'. Check the filename and folder.")
else:
    # Step 1: Resize to width=200 and height=300
    resized_product = cv2.resize(
        product,
        (200, 300),
        interpolation=cv2.INTER_AREA
    )

    # Step 2: Flip the resized image vertically (flipCode=0)
    transformed_product = cv2.flip(resized_product, 0)

    cv2.imwrite("myntra_transformed.jpg", transformed_product)
    print("Resized dimensions (height, width):", resized_product.shape[:2])
    print("Saved final transformed image as myntra_transformed.jpg")

    plt.figure(figsize=(8, 5))
    plt.imshow(cv2.cvtColor(transformed_product, cv2.COLOR_BGR2RGB))
    plt.title("Myntra Product — Resized and Vertically Flipped")
    plt.axis("off")
    plt.show()
```
